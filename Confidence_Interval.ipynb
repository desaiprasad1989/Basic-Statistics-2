##### a. Build 99% Confidence Interval Using Sample Standard Deviation Assuming the sample is representative of the population, construct a 99% confidence interval for the mean number of characters printed before the print-head fails using the sample standard deviation. Explain the steps you take and the rationale behind using the t-distribution for this task.

```python
# Scipy for Stats Function Data and Pandas and Numpy for Data Manupulation
from scipy import stats
import math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### Get Sample Data

```python
df = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
```

```python
df = pd.DataFrame(df)
df
```

```python
df.rename({0:'Sample'}, axis=1, inplace=True)
```

```python
df['Sample'].mean()
```

```python
df['Sample'].std()
```

```python
df.describe()
```

```python
# Confidnce Interval = (n, loc/Mean, SD/Scale)

sample_mean = df['Sample'].mean()
sample_std_dev = df['Sample'].std()
sample_size = df['Sample'].count()
confidence_level = 0.99

## Degrees of freedom
dof = sample_size - 1

## Significance Level
alpha = (1 - confidence_level) / 2

## finding t- value from t distibution table
t_value = stats.t.ppf(1 - alpha, dof)

margin_of_error = t_value * (sample_std_dev / np.sqrt(sample_size))

lower_limit = sample_mean - margin_of_error
upper_limit = sample_mean + margin_of_error

print("dof:", dof)
print("alpha:", alpha)
print("t_value:", t_value)
print("margin_of_error:", margin_of_error)

print(f"99% Confidence Interval for the mean number of characters printed before the print-head fails using sample Standard Deviation :")
print(f"Confidence Interval : ({lower_limit}, {upper_limit})")
```

### Observtion :

99% confidence interval ranges between (1.0901973384384906, 1.3871359948948425) for the mean number of characters printed before the print-head fails using the sample standard deviation.
Using t test, cosidering the sample as population and the standard deviation is not known.

##### b. Build 99% Confidence Interval Using Known Population Standard Deviation If it were known that the population standard deviation is 0.2 million characters, construct a 99% confidence interval for the mean number of characters printed before failure.

```python
## confidence Interval = xbar +- z * (sigma / sqrt(n))

## left_limit = xbar - Z * (sigma / sqrt(n))
## right_limit = xbar  Z * (sigma / sqrt(n))

Z value for 99% confidence level from Z table
```

```python
xbar = df['Sample'].mean()
Z = 2.58
pop_std_dev = 0.2
n = len(df['Sample'])

left_limit = xbar - ( Z * (pop_std_dev / np.sqrt(n)))
right_limit = xbar + ( Z * (pop_std_dev / np.sqrt(n)))

print(f"99% confidence interval for the mean number of characters printed before failure for known population standard deviation 0.2 is :")
print(f"Confidence Interval : ({left_limit}, {right_limit})")
```

### Observation : 
The confidence interval with 99% with known standard deviation is (1.1054360395571314, 1.3718972937762017)

### Conclusion :
#### We can conclude that the confidence interval with 99%, with unknown population standard deviation using t test and with known population standard deviation at 0.2 using Z test is approximatetly equal.
